```python

import numpy as np
import csv

# Names of the five counties
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

# Months of the year
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

# Rainfall data for Nairobi, Nakuru, Kisumu and Mombasa
clean_rows = [
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70]
]

# Raw rainfall data for Machakos
machakos_raw = [
    "48", "35", "88", "N/A", "140", " 20 ",
    "9", "12", "", "45", "182", "96"
]

print("Counties:", counties)
print("Number of counties:", len(counties))
print("Number of months:", len(months))
print("Rainfall data for four counties has been entered.")
print("Machakos raw data:", machakos_raw)
```

```python

def clean_machakos(raw_values):
    values = []
    invalid_indices = []

    for i, value in enumerate(raw_values):
        try:
            # Convert valid text readings into integers
            values.append(int(value.strip()))
        except ValueError:
            # Record invalid readings for replacement
            values.append(None)
            invalid_indices.append(i)

    # Calculate the average of valid readings only
    valid_values = [v for v in values if v is not None]
    average = round(sum(valid_values) / len(valid_values))

    # Replace invalid readings with the rounded average
    for i in invalid_indices:
        values[i] = average

    return values, invalid_indices


# Clean the Machakos rainfall data
machakos, imputed_indices = clean_machakos(machakos_raw)

# Print the results
print("Cleaned Machakos data:", machakos)
print("Imputed months:", [months[i] for i in imputed_indices])
print("Imputed rainfall value:", machakos[imputed_indices[0]], "mm")
```

```python

# Combine the four clean rows and the cleaned Machakos row
R = np.array(clean_rows + [machakos])

# Print the complete rainfall array
print("Combined rainfall array:")
print(R)

# Print the shape and data type
print("Array shape:", R.shape)
print("Array dtype:", R.dtype)
```

```python

# Calculate total rainfall for each county
annual_totals = np.sum(R, axis=1)

print("ANNUAL RAINFALL TOTALS")
for county, total in zip(counties, annual_totals):
    print(f"{county}: {total} mm")


# Calculate average rainfall for each month across all five counties
monthly_averages = np.mean(R, axis=0)

print("\nAVERAGE MONTHLY RAINFALL")
for month, average in zip(months, monthly_averages):
    print(f"{month}: {average:.2f} mm")
```

```python

# Find the index of the highest rainfall value in each county
wettest_indices = np.argmax(R, axis=1)

# Convert each index into its month name
wettest_months = [months[i] for i in wettest_indices]

# Print each county and its wettest month using zip()
print("WETTEST MONTH FOR EACH COUNTY")

for county, wettest_month in zip(counties, wettest_months):
    print(f"{county}: {wettest_month}")
```

```python

# Create a Boolean mask: True where rainfall is below 50 mm
dry_mask = R < 50

print("Boolean mask for dry months:")
print(dry_mask)

# Count dry months for each county
dry_month_counts = np.sum(dry_mask, axis=1)

# Print each county's dry-month count
print("\nNUMBER OF DRY MONTHS")

for county, count in zip(counties, dry_month_counts):
    print(f"{county}: {count} dry months")
```
